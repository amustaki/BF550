```python
class random_walker(base_walker):
    def __init__(self):
        # define stored variables, initial data
        self.x = 0
        self.walk = [self.x]
    def one_step(self):
        # increment the current position by +1 or -1, with equal probability
        rand = random.random() #chooses a random value between 0 and 1 
        #do bins to break up 1 into thirds to do the probability 
        if rand < 0.33: #if its less 0.3 it moves to the right
            self.x += 1
        if rand > 0.33 and rand < 0.66:
            self.x -= 1 #if its greater than 0.3 and less than 0.66 it moves to the left  
        else: 
            self.x += 0 #else it stays the same 
        self.walk.append(self.x)
    def get_walk(self):
        # return the history of the position of the walker
        return self.walk
```
